### Pandas Practise of DataFrame 

```python
import numpy as np
import pandas as pd
```

### Creating a DataFrame

```python
data={
    "Name":['John','Aman','Naman','Vikas'],
    "Age":[20,21,19,22],
    'City':['Bhopal','Indore','Reva','Nagpur'],
    'Salary':[10000,25000,15000,20000]
}
```

```python
df=pd.DataFrame(data)
df
```

### Create dataframe using List

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
```

```python
df2=pd.DataFrame(data_list,)
df2
```

```python
columns=["Name",'Age','City',"Salary"]
df3=pd.DataFrame(data_list,columns=columns)
df3
```
