## ***** -------------- PANDAS ----------------- *****

##### Pandas (stands for Python Data Analysis) is an open-source software library designed for data manipulation and analysis.

```python
import numpy as np
import pandas as pd
print("Import pandas successfully")
```

```python
labels=['a','b','c']
mlist=[10,20,40]
arr=np.array([10,20,30])
d={1:10,2:20,3:30}
```

#### SERIES using list

```python
pd.Series(mlist)
```

#### SERIES using list and labels

```python
pd.Series(mlist,index=labels)
```

#### SERIES using Array

```python
pd.Series(arr,index=labels)
```

#### SERIES using Dictionary

```python
pd.Series(d)
```
