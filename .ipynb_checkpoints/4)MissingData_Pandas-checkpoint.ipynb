## ********* Handling Missing Data in Pandas  ********* 

```python
import numpy as np
import pandas as pd
print("Import Successfully")
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
```

```python
data
```

```python
df=pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

### Remove Missing Data

```python
df
```

```python
df.dropna()
```

```python
df
```

```python
df.dropna(thresh=3)
```

```python
df.dropna(thresh=1)
```

### Filling  the Missing Value

```python
df
```

```python
df.fillna(0)
```

```python
df
```

```python
values={'A':0,'B':100,'C':300,'D':400}
df.fillna(value=values)
```

```python
df.fillna(df.mean())
```
