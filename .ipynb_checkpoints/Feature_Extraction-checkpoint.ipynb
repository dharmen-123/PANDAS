## *********** FEATURES EXTRACTION *********** 

```python
# import Libraries
import numpy as np
import pandas as pd
```

```python
# load the Dataset
df=pd.read_csv(r'anime.csv')
df
```

```python
df.head()
```

### make the new Column for episode count

```python
df.loc[3]['Title']
```

```python
def extract_episodes(text):
    epi=''
    check=False
    for i in text:
        if(i==')'):
            break
        if(i=='('):
            check=True
        elif(check):
            epi+=i
    return epi
```

```python
df['Episodes']=df['Title'].apply(extract_episodes)
```

```python
df.head()
```

```python
df['Episodes']=df['Episodes'].str.replace(" eps","")
df.head()
```

```python
df['Episodes'].dtype
```

```python
# Change the data type of Episodes
df['Episodes']=df['Episodes'].astype(int)
df['Episodes'].dtype
```

### Make a new column for Time Stamp

```python
df.loc[1]['Title']
```

```python
def timestamp_extraction(text):
    check=False
    t=''
    j=1
    for i in text:
        if j==20:
            break
        if i==')':
            check=True
        elif check:
            t=t+i
            j=j+1        
    return t
```

```python
df['Time Stamp']=df['Title'].apply(timestamp_extraction)
```

```python
df.head()
```

### Make a new column of Months

```python
# Function to calculate months
def calculate_months(period):
    start, end = period.split(' - ')
    start = pd.to_datetime(start, format='%b %Y')
    end = pd.to_datetime(end, format='%b %Y')

    months = (end.year - start.year) * 12 + (end.month - start.month)+1 # +1 include the staring month
    return months
```

```python
df['Months']=df['Time Stamp'].apply(calculate_months)
df.head()
```
